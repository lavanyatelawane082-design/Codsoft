```python
# Import Libraries
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
```

```python
# Loading Dataset
```

```python
df = pd.read_csv("creditcard.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.describe()
```

```python
# Check how many fraud and normal transactions
```

```python
print("\nClass distribution:")
print(df['Class'].value_counts())
```

```python
sns.countplot(x=df['Class'])
plt.title("0 = Normal, 1 = Fraud")
plt.show()
```

```python
# Check missing values
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
scaler = StandardScaler()
df['Amount'] = scaler.fit_transform(df[['Amount']])
```

```python
df = df.drop('Time', axis=1)
```

```python
# Handle Class Imbalance
```

```python
fraud = df[df['Class'] == 1]
normal = df[df['Class'] == 0]

print("\nFraud Cases:", len(fraud))
print("Normal Cases:", len(normal))
```

```python
# Take 3x normal samples
```

```python
normal_sample = normal.sample(n=len(fraud)*3, random_state=42)
```

```python
data = pd.concat([fraud, normal_sample])
data = data.sample(frac=1, random_state=42)
print("\nBalanced data:")
print(data['Class'].value_counts())
```

```python
# Preparing data for model
```

```python
x = data.drop('Class', axis=1)
y= data['Class']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# Training the model
```

```python
model = RandomForestClassifier(n_estimators = 100, random_state = 42)
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
print("\nmodel Results:\n")
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

sns.heatmap(cm, annot=True)
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```
